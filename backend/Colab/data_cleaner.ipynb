```python
#DATA CLEANING

phishing_emails = "/content/drive/MyDrive/SevenPhishingEmails/"

dataframes = []

for file in os.listdir(phishing_emails):
    #print(file)
    if file.endswith('.csv'):
        #print(f"{file} ENDS WITH CSV!")
        file_path = os.path.join(phishing_emails, file)

        df = pd.read_csv(file_path, on_bad_lines='skip')

        dataframes.append(df)

combined_df = pd.concat(dataframes,ignore_index=True)
combined_df.drop_duplicates(inplace=True)
combined_df = combined_df.fillna('')

cleaned_file_path = os.path.join(phishing_emails, "scikit_cleaned.csv")
combined_df.to_csv(cleaned_file_path, index=False)

print(f"Data has been cleaned to {cleaned_file_path} \n DATA --> \n {combined_df.info()} \n MISSING --> \n {combined_df.isnull().sum()}")
```
